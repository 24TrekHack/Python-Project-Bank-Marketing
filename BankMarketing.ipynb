Import Python Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Data Analyses

```python
df = pd.read_csv("bank-additional-full.csv", sep=";")

print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns)

print("\nFirst 5 rows:")
print(df.head())

print("\nMissing Values:")
print(df.isnull().sum())
```

Data Visualization

```python
subscription_counts = df["y"].value_counts().reindex(["no", "yes"], fill_value=0)

plt.figure(figsize=(8, 5))
plt.bar(["Did not subscribe", "Subscribed"], subscription_counts.values)
plt.xlabel("Subscription Outcome")
plt.ylabel("Number of Customers")
plt.title("Subscription Outcome")
plt.savefig("subscription_outcome_bar_chart.png")
plt.show()

plt.figure(figsize=(8, 5))
plt.hist(df["age"].dropna(), bins=20, edgecolor="black")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.title("Distribution of Customer Ages")
plt.savefig("customer_age_histogram.png")
plt.show()

job_rate = df.groupby("job")["y"].apply(lambda values: (values == "yes").mean() * 100).sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(job_rate.index, job_rate.values)
plt.xlabel("Job Category")
plt.ylabel("Subscription Rate (%)")
plt.title("Subscription Rate by Job")
plt.xticks(rotation=35)
plt.savefig("subscription_rate_by_job.png")
plt.show()

month_order = ["jan", "feb", "mar", "apr", "may", "jun", "jul", "aug", "sep", "oct", "nov", "dec"]
monthly_contacts = df["month"].value_counts().reindex(month_order, fill_value=0)

plt.figure(figsize=(9, 5))
plt.bar([month.title() for month in monthly_contacts.index], monthly_contacts.values)
plt.xlabel("Month")
plt.ylabel("Number of Records")
plt.title("Number of Campaign Records by Month")
plt.savefig("contacts_by_month.png")
plt.show()

contact_order = [method for method in ["cellular", "telephone"] if method in df["contact"].unique()]
contact_rate = df.groupby("contact")["y"].apply(lambda values: (values == "yes").mean() * 100).reindex(contact_order)

plt.figure(figsize=(8, 5))
plt.bar(contact_rate.index, contact_rate.values)
plt.xlabel("Contact Method")
plt.ylabel("Subscription Rate (%)")
plt.title("Subscription Rate by Contact Method")
plt.savefig("subscription_rate_by_contact_method.png")
plt.show()

previous_order = ["success", "failure", "nonexistent"]
previous_rate = df.groupby("poutcome")["y"].apply(lambda values: (values == "yes").mean() * 100).reindex(previous_order)
previous_labels = ["Success", "Failure", "No previous outcome"]

plt.figure(figsize=(8, 5))
plt.bar(previous_labels, previous_rate.values)
plt.xlabel("Previous Campaign Outcome")
plt.ylabel("Subscription Rate (%)")
plt.title("Subscription Rate by Previous Campaign Outcome")
plt.savefig("subscription_rate_by_previous_outcome.png")
plt.show()

no_subscription_duration = df.loc[df["y"] == "no", "duration"].dropna()
yes_subscription_duration = df.loc[df["y"] == "yes", "duration"].dropna()

plt.figure(figsize=(8, 5))
plt.boxplot([no_subscription_duration, yes_subscription_duration])
plt.xticks([1, 2], ["Did not subscribe", "Subscribed"])
plt.xlabel("Subscription Outcome")
plt.ylabel("Call Duration (seconds)")
plt.title("Call Duration by Subscription Outcome")
plt.savefig("call_duration_box_plot.png")
plt.show()

campaign_order = [str(number) for number in range(1, 11)] + ["11+"]
campaign_groups = np.where(df["campaign"] >= 11, "11+", df["campaign"].astype(str))
df["campaign_group"] = campaign_groups
campaign_rate = df.groupby("campaign_group")["y"].apply(lambda values: (values == "yes").mean() * 100).reindex(campaign_order)

plt.figure(figsize=(10, 5))
plt.bar(campaign_rate.index, campaign_rate.values)
plt.xlabel("Number of Contacts")
plt.ylabel("Subscription Rate (%)")
plt.title("Subscription Rate by Number of Campaign Contacts")
plt.savefig("subscription_rate_by_campaign_contacts.png")
plt.show()
```

Conclusion:

The analysis provides an overview of the bank marketing campaign and customer subscription outcomes. The visualizations show patterns related to customer age, job, contact method, previous campaign outcome, call duration, and number of contacts. These insights can help understand factors associated with term-deposit subscriptions.
